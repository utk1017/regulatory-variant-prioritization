## Inspect the provenance manifest

Every run writes `manifest.json` alongside its output — this is what makes a result traceable back
to the exact ClinVar snapshot, reference build, and config that produced it.

```
import json

gene_symbol = "HBB"  # matches configs/hbb.yaml
manifest_path = f"data/outputs/{gene_symbol}/manifest.json"
with open(manifest_path) as f:
    manifest = json.load(f)

manifest
```
